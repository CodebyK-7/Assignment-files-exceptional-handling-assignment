Q1: What is the difference between interpreted and compiled languages?

A1: Interpreted languages execute code line by line at runtime, while compiled languages translate code into machine code before execution.



Q2: What is exception handling in Python?

A2: Exception handling allows you to manage runtime errors gracefully using try, except, else, and finally blocks.



Q3: What is the purpose of the finally block in exception handling?

A3: The finally block ensures cleanup code (like closing files or releasing resources) always runs, regardless of exceptions.



Q4: What is logging in Python?

A4: Logging records events during program execution, useful for debugging and monitoring.

Q5: What is the significance of the __del__ method in Python?

A5: The __del__ method is a destructor called when an object is about to be destroyed, often used for cleanup.

Q6: What is the difference between import and from ... import in Python?
A6: import loads the entire module, while from ... import loads specific attributes or functions.

Q7: How can you handle multiple exceptions in Python?

A7: By listing them in a tuple inside the except block, e.g., except (ValueError, TypeError):

Q8: What is the purpose of the with statement when handling files in Python?

A8: It ensures proper resource management by automatically closing files after use.

Q9: What is the difference between multithreading and multiprocessing?
A9: Multithreading runs multiple threads within a process (sharing memory), while multiprocessing runs separate processes (independent memory).

Q10: What are the advantages of using logging in a program?

A10: Logging provides persistent records, helps debug issues, and allows monitoring of application behavior.

Q11: What is memory management in Python?

A11: Memory management involves allocation, deallocation, and garbage collection of objects.

Q12: What are the basic steps involved in exception handling in Python?
A12: Use try for risky code, except for handling errors, else for code if no error, and finally for cleanup.

Q13: Why is memory management important in Python?

A13: It prevents memory leaks and ensures efficient resource usage.

Q14: What is the role of try and except in exception handling?

A14: try defines code that may raise exceptions, except handles those exceptions.

Q15: How does Python's garbage collection system work?

A15: It frees memory by removing objects with zero references, using reference counting and cyclic garbage collection.

Q16: What is the purpose of the else block in exception handling?

A16: The else block runs if no exception occurs in the try block.

Q17: What are the common logging levels in Python?

A17: DEBUG, INFO, WARNING, ERROR, CRITICAL.

Q18: What is the difference between os.fork() and multiprocessing in Python?

A18: os.fork() creates a child process at the OS level (Unix only), while multiprocessing is cross-platform and provides higher-level process management.

Q19: What is the importance of closing a file in Python?

A19: Closing a file ensures data is saved and resources are released.

Q20: What is the difference between file.read() and file.readline() in Python?

A20: file.read() reads the entire file or specified bytes, while file.readline() reads one line at a time.

Q21: What is the logging module in Python used for?

A21: It provides a flexible framework for emitting log messages to files, consoles, or other outputs.

Q22: What is the os module in Python used for in file handling?

A22: It helps with tasks like creating, deleting, and navigating directories.

Q23: What are the challenges associated with memory management in Python?

A23: Challenges include memory leaks, handling large objects, and managing cyclic references.

Q24: How do you raise an exception manually in Python?

A24: Use the raise keyword, e.g., raise ValueError('Invalid input').

Q25: Why is it important to use multithreading in certain applications?

A25: Multithreading improves performance in I/O-bound tasks by enabling concurrency.

```python
# Q2: Write a Python program to read the contents of a file and print each line.
with open("example.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
# Q3: How would you handle a case where the file doesn’t exist while trying to open it for reading?
try:
    with open("nofile.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found!")
```

```python
# Q4: Write a Python script that reads from one file and writes its content to another file.
try:
    with open("source.txt", "r") as src, open("dest.txt", "w") as dst:
        dst.write(src.read())
except FileNotFoundError:
    print("Source file not found!")
```

```python
# Q5: How would you catch and handle division by zero error in Python?
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Division by zero error!")
```

```python
# Q6: Write a Python program that logs an error message to a log file when a division by zero exception occurs.
import logging
logging.basicConfig(filename="errors.log", level=logging.ERROR)
try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Division by zero occurred!")
```

```python
# Q7: How do you log information at different levels (INFO, ERROR, WARNING) in Python using the logging module?
import logging
logging.basicConfig(level=logging.DEBUG)
logging.info("This is info")
logging.warning("This is warning")
logging.error("This is error")
```

```python
# Q11: Write a Python program that uses a try–except block to handle an error when attempting to access a dictionary key that doesn’t exist.
data = {"name": "Kritika"}
try:
    print(data["age"])
except KeyError:
    print("Key not found!")
```

```python
# Q12: Write a program that demonstrates using multiple except blocks to handle different types of exceptions.
try:
    num = int("abc")
    result = 10 / 0
except ValueError:
    print("ValueError occurred")
except ZeroDivisionError:
    print("ZeroDivisionError occurred")
```

```python
# Q13: How would you check if a file exists before attempting to read it in Python?
import os
if os.path.exists("example.txt"):
    with open("example.txt", "r") as f:
        print(f.read())
else:
    print("File does not exist")
```

```python
# Q17: Write a Python program to create and write a list of numbers to a file, one number per line.
numbers = [1, 2, 3, 4, 5]
with open("numbers.txt", "w") as f:
    for num in numbers:
        f.write(str(num) + "\n")
```

```python
# Q18: How would you implement a basic logging setup that logs to a file with rotation after 1MB?
import logging
from logging.handlers import RotatingFileHandler
handler = RotatingFileHandler("rotating.log", maxBytes=1024*1024, backupCount=3)
logging.basicConfig(handlers=[handler], level=logging.INFO)
logging.info("This is a rotating log example")
```

```python
# Q19: Write a program that handles both IndexError and KeyError using a try–except block.
try:
    lst = [1, 2, 3]
    print(lst[5])
    d = {"a": 1}
    print(d["b"])
except IndexError:
    print("IndexError occurred")
except KeyError:
    print("KeyError occurred")
```

```python
# Q21: Write a Python program that reads a file and prints the number of occurrences of a specific word.
word = "Python"
with open("example.txt", "r") as f:
    content = f.read()
print(content.count(word))
```

```python
# Q22: How can you check if a file is empty before attempting to read its contents?
import os
if os.path.getsize("example.txt") == 0:
    print("File is empty")
else:
    with open("example.txt", "r") as f:
        print(f.read())
```

```python
# Q23: Write a Python program that writes to a log file when an error occurs during file handling.
import logging
logging.basicConfig(filename="file_errors.log", level=logging.ERROR)
try:
    with open("nofile.txt", "r") as f:
        print(f.read())
except Exception as e:
    logging.error("Error occurred: %s", e)
```
